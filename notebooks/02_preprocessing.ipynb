```python
from bankiq.features import clean_narration, TransactionFeaturizer, NUMERIC_COLS, numeric_features, month_panel, customer_profile, PANEL_FEATURES, PROFILE_FEATURES
from bankiq.data.generator import load
tx, cu = load()
for s in tx['narration'].sample(5, random_state=1):
    print(f'{s!r:70} -> {clean_narration(s)!r}')
```

```python
fz = TransactionFeaturizer().fit(tx.sample(20000, random_state=0))
X = fz.transform(tx.head(5))
print('features:', fz.n_features, '| word:', len(fz.word.vocabulary_), '| char:', len(fz.char.vocabulary_), '| numeric:', len(NUMERIC_COLS))
X
```
