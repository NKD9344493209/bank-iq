# Bank IQ · 07 — The intelligence service on a real statement

Runs the trained models on one of the sample bank statements exactly as the `/api/ml/analyze` endpoint does: ML categories vs regex rules, monthly panel, segment, forecast, anomalies and the generated business insights.

> Run from the `notebooks/` folder with the repo root on `sys.path` (first cell does this). Figures are written to `results/figures/`, metrics to `results/*.json`, models to `models/`.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
os.chdir('..') if os.path.basename(os.getcwd()) == 'notebooks' else None
%matplotlib inline
```

```python
from pathlib import Path
from vasool.parsers import parse_file
from bankiq import service
p = Path('data/samples/hdfc_priya_salary_2026.csv')
parsed = parse_file(p)
rows = [{'date': r['date'], 'narration': r['narration'], 'debit': r.get('debit', 0), 'credit': r.get('credit', 0), 'balance': r.get('balance')} for r in parsed.rows]
out = service.analyze(rows, p.stem)
out['totals'], out['segment'], out['forecast']
```
